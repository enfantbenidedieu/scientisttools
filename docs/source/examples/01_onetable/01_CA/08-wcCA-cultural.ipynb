## cultural dataset

```python
#children dataset
from scientisttools.datasets import cultural
cultural
```

### Instanciation and training

```python
# instanciation
from scientisttools import CA
clf = CA(
    group = 0,
    option = "within",
    row_sup = range(20,26),
    col_sup = range(9,12),
    sup_var = range(12,18)
)
```

### `fit` function

```python
# fit function
clf.fit(cultural)
```

### `fit_transform` function

```python
#fit_transform - rows coordinates
row_coord = clf.fit_transform(cultural)
row_coord
```

### `transform` function

```python
#transform function - rows coordinates
#row_coord = clf.transform(clf.call_.X)
#row_coord
```

## Eigenvalues

```python
#eigen values
from scientisttools import get_eig, fviz_eig
eig = get_eig(clf)
eig
```

```python
# scree plot - eigenvalue
p = fviz_eig(
    obj = clf, 
    choice = "eigenvalue", 
    show_labels = True
)
print(p.show())
```

```python
# scree plot - proportion of variance
p = fviz_eig(
    obj = clf,
    choice = "proportion",
    show_labels = True
)
print(p.show())
```

## Rows informations

```python
#rows informations
from scientisttools import get_ca, fviz_ca
row = get_ca(clf,element = "row")
row._fields
```
