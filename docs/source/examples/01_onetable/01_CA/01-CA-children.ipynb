## children dataset

```python
#children dataset
from scientisttools.datasets import children
children.data
```

### Instanciation and training

```python
# instanciation
from scientisttools import CA
# with supplementary columns
clf = CA(
    row_sup = range(14,18),
    col_sup = (5,6,7),
    sup_var = 8
)
# with supplementary variables
clf2 = CA(
    row_sup = range(14,18),
    sup_var = range(5,9)
)
```

### `fit` function

```python
# fit function - with supplementary columns
clf.fit(children.data)
```

```python
# fit function - with supplementary quantitative variables
clf2.fit(children.data)
```

### `fit_transform` function

```python
#fit_transform - rows coordinates
row_coord = clf.fit_transform(children.data)
row_coord
```

### `transform` function

```python
#transform function - rows coordinates
row_coord = clf.transform(clf.call_.X)
row_coord
```

## Eigenvalues

```python
#eigen values
from scientisttools import get_eig, fviz_eig
eig = get_eig(clf)
eig
```

```python
# scree plot - eigenvalue
p = fviz_eig(
    obj = clf, 
    choice = "eigenvalue", 
    show_labels = True
)
print(p.show())
```

```python
# scree plot - proportion of variance
p = fviz_eig(
    obj = clf,
    choice = "proportion",
    show_labels = True
)
print(p.show())
```
