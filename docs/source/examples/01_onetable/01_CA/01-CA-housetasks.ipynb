## housetasks dataset

```python
#housetasks dataset
from scientisttools.datasets import housetasks
housetasks
```

### Instanciation and training

```python
# instanciation
from scientisttools import CA
clf = CA()
```

### `fit` function

```python
# fit function
clf.fit(housetasks)
```

### `fit_transform` function

```python
# fit_transform function - rows coordinates
row_coord = clf.fit_transform(housetasks)
row_coord
```

### `transform` function

```python
# transform function - rows coordinates
row_coord = clf.transform(clf.call_.X)
row_coord
```

## Eigenvalues

```python
#eigen values
from scientisttools import get_eig, fviz_eig
eig = get_eig(clf)
eig
```

```python
# scree plot - eigenvalue
p = fviz_eig(
    obj = clf,
    choice = "eigenvalue",
    show_labels = True
)
print(p.show())
```

```python
# scree plot - proportion of variance
p = fviz_eig(
    obj = clf,
    choice = "proportion",
    show_labels = True
)
print(p.show())
```

## Row informations

```python
# rows informations
from scientisttools import get_ca, fviz_ca
row = get_ca(clf,element = "row")
row._fields
```

### Rows coordinates

```python
# rows coordinates
row.coord
```

```python
# row points
p = fviz_ca(
    obj = clf, 
    choice = "row",
    repel = True
)
print(p.show()) 
```

```python
# rows contributions
row.contrib
```

## Statistics with CA 

```python
# statistics with CA
from scientisttools import statsCA
stats = statsCA(clf)
stats._fields
```

### Diagnostics tests - multivariate goodness of fit tests

```python
# multivariate goodness of fit
goodness = stats.goodness
goodness._fields
```

#### Pearson chi-squared test and Log-likelihood test (G-test)

```python
# pearson chi-squared test and Log-likelihood test (G-test)
goodness.test
```

#### Association measures

```python
# association measures
goodness.association
```
