## ichtyo dataset

```python
#housetasks dataset
from scientisttools.datasets import ichtyo
ichtyo
```

### Instanciation and training

```python
# instanciation
from scientisttools import CA
clf = CA(ref=9,sup_var=10)
```

### `fit` function

```python
# fit function
clf.fit(ichtyo)
```

### `fit_transform` function

```python
# fit_transform function - rows coordinates
row_coord = clf.fit_transform(ichtyo)
row_coord
```

### `transform` function

```python
# transform function - rows coordinates
row_coord = clf.transform(ichtyo.iloc[:,:10])
row_coord
```

## Eigenvalues

```python
#eigen values
from scientisttools import get_eig, fviz_eig
eig = get_eig(clf)
eig
```

```python
# scree plot - eigenvalue
p = fviz_eig(
    obj = clf,
    choice = "eigenvalue",
    show_labels = True
)
print(p.show())
```

```python
# scree plot - proportion of variance
p = fviz_eig(
    obj = clf,
    choice = "proportion",
    show_labels = True
)
print(p.show())
```
