## housetasks dataset

```python
#housetasks dataset
from scientisttools.datasets import housetasks
housetasks
```

### Instanciation and training

```python
# instanciation
from scientisttools import CA
clf = CA(symmetric=False)
```

### `fit` function

```python
# fit function
clf.fit(housetasks)
```

### `fit_transform` function

```python
# fit_transform function - rows coordinates
row_coord = clf.fit_transform(housetasks)
row_coord
```

### `transform` function

```python
# transform function - rows coordinates
row_coord = clf.transform(housetasks)
row_coord
```

## Eigenvalues

```python
#eigen values
from scientisttools import get_eig, fviz_eig
eig = get_eig(clf)
eig
```

```python
# scree plot - eigenvalue
p = fviz_eig(
    obj = clf,
    choice = "eigenvalue",
    show_labels = True
)
print(p.show())
```

```python
# scree plot - proportion of variance
p = fviz_eig(
    obj = clf,
    choice = "proportion",
    show_labels = True
)
print(p.show())
```

## Row informations

```python
# rows informations
from scientisttools import get_ca, fviz_ca
row = get_ca(clf,element = "row")
row._fields
```

### Rows coordinates

```python
# rows coordinates
row.coord
```

```python
# row points
p = fviz_ca(
    obj = clf, 
    choice = "row",
    repel = True
)
print(p.show()) 
```

### Rows contributions

```python
# rows contributions
row.contrib
```

```python
# row points colored by contrib
p = fviz_ca(
    obj = clf, 
    choice = "row",
    col_row = "contrib",
    repel = True
)
print(p.show())
```

### Rows cos2

```python
# rows cos2
row.cos2
```

```python
# row points colored by cos2
p = fviz_ca(
    obj = clf, 
    choice = "row", 
    col_row = "cos2",
    repel = True
)
print(p.show())
```

### Rows additionals informations

```python
# rows additionals informations
row.infos
```
