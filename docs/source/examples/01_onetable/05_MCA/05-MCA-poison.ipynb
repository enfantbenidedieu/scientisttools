### Data reconstruction - `reconst`

```python
#disjunctive table reconstitution
from scientisttools import reconst
rec_disj = reconst(clf,clf.call_.ncp-1)
rec_disj
```

```python
# data reconstitution
from pandas import from_dummies, concat
rec = from_dummies(rec_disj,sep="_")
# mapping
rec_map = concat((rec[k].map({x : f"{k}_{x}" for x in rec[k].unique()}) for k in rec.columns),axis=1)
rec_map
```

```python
# confusion matrix
from pandas import crosstab
for k,l in zip(clf.call_.X.columns,rec_map.columns):
    print(f"(Rows = {k} ; Columns = {l})")
    print(crosstab(clf.call_.X[k],rec_map[l]))
    print("")
```
