## tea dataset

```python
# tea dataset
from scientisttools.datasets import tea
data = tea.iloc[:,:22]
data.info()
```

## Instanciation and training

```python
# instanciation & training
from scientisttools import MCA
clf = MCA(iv=(18,19,20,21),ortho=True)
```

### `fit` function

```python
#fit function
clf.fit(data)
```

### `fit_transform` function

```python
#fit_transform - individuals coordinates
ind_coord = clf.fit_transform(data)
ind_coord
```

### `transform` function

```python
# transform - individuals coordinates
ind_coord = clf.transform(data)
ind_coord
```

```python
# attributes
from scientisttools import sprintf
sprintf(clf)
```

## Eigen values

```python
#eigenvalues
from scientisttools import get_eig
eig = get_eig(clf)
eig
```

```python
# scree plot - eigenvalue
from scientisttools import fviz_screeplot
p = fviz_screeplot(obj=clf,choice="eigenvalue")
print(p.show())
```

```python
# scree plot - proportion
p = fviz_screeplot(obj=clf,choice="proportion")
print(p.show())
```

```python
# scree plot - cumulative
p = fviz_screeplot(obj=clf,choice="cumulative")
print(p.show())
```

## Individuals informations

```python
# individuals informations
from scientisttools import get_mca
ind = get_mca(clf,"ind")
ind._fields
```

### Individuals coordinates

```python
# individuals coordinates
ind.coord
```

```python
# graph of individuals
from scientisttools import fviz_mca
p = fviz_mca(
    obj=clf,
    choice="ind",
    repel = True,
    ind_sup = False
)
print(p.show())
```
