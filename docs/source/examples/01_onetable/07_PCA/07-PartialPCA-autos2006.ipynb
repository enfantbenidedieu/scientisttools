## autos2006 dataset

```python
# autos2006 dataset
from scientisttools.datasets import autos2006
data = autos2006.data
data.info()
```

## Instanciation & training

```python
# instanciation
from scientisttools import PCA
clf = PCA(partial=0,ind_sup=(18,19),sup_var=(6,7,8))
```

### `fit` function

```python
# fit function
clf.fit(data)
```

### `fit_transform` function

```python
# fit_transform function - individuals coordinates
ind_coord = clf.fit_transform(data)
ind_coord
```

### `transform` function

```python
# transform function - individuals coordinates
ind_coord = clf.transform(autos2006.actif)
ind_coord
```

```python
# attributes
from scientisttools import sprintf
sprintf(clf)
```
