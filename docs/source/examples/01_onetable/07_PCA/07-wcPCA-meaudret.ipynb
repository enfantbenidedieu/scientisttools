# Within-Class Analysis - PCA - meaudret dataset

```python
#disable warnings
from warnings import simplefilter, filterwarnings
simplefilter(action='ignore', category=FutureWarning)
filterwarnings("ignore")
```

## meauret dataset

```python
# meaudret dataset
from scientisttools.datasets import meaudret
meaudret._fields
```

```python
# overall dataset
data = meaudret.data
data.info()
```

## Instanciation and training

```python
# instanciation
from scientisttools import PCA
clf = PCA(group=9,option="within",sup_var=range(10,24))
```

### `fit` function

```python
# fit function
clf.fit(data)
```

### `fit_transform` function

```python
# fit_transform - individuals coordinates
ind_coord = clf.fit_transform(data)
ind_coord
```

### `transform` function

```python
# transform - individuals coordinates
ind_coord = clf.transform(meaudret.actif)
ind_coord
```

```python
# attributes
from scientisttools import sprintf
sprintf(clf)
```

## Eigen values

```python
# eigenvalues
from scientisttools import get_eig
eig = get_eig(clf)
eig
```

```python
# scree plot
from scientisttools import fviz_screeplot
p = fviz_screeplot(clf,choice="eigenvalue")
print(p.show())
```

```python
# proportion
p = fviz_screeplot(clf,choice="proportion")
print(p.show())
```

```python
# cumulative proportion
p = fviz_screeplot(clf,choice="cumulative")
print(p.show())
```

## Individuals informations

```python
# individuals informations
from scientisttools import get_pca
ind = get_pca(clf,"ind")
ind._fields
```

### Individuals coordinates

```python
# individuals coordinates
ind.coord
```

```python
# graph of individuals
from scientisttools import fviz_pca
p = fviz_pca(
    obj=clf,
    choice="ind",
    repel=True,
    ind_sup=False,
    quali_sup=False
)
print(p.show())
```

### Individuals cos2

```python
# individuals cos2
ind.cos2
```

```python
# graph of individuals - color by cosinus
p = fviz_pca(
    obj=clf,
    choice="ind",
    col_ind="cos2",
    repel=True,
    ind_sup=False,
    quali_sup=False
)
print(p.show())
```

```python
# graph of cosinus
from scientisttools import fviz_corrplot
p = fviz_corrplot(
    X=ind.cos2,
    legend_title="Cos2",
    title="Cosinus of individuals"
)
print(p.show())
```

```python
# cosinus of individuals to axis 1
from scientisttools import fviz_cos2
p = fviz_cos2(
    obj=clf,
    choice="ind"
)
print(p.show())
```
